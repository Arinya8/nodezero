# Rule Engine MVP — Interactive Demonstration Notebook

This notebook demonstrates the executable, deterministic packaging recommendation engine built according to `rule_engine_mvp_plan.md`.

## Engine Architecture
1. **Group 1 Regulatory Rules (R1.1 – R1.6)**: Evaluates material prohibitions (Pan Masala), recycled plastics, rPET decontamination proof, thickness floors, migration limits, and ink safety.
2. **Group 2 Physics / Barrier Rules (R2.1 – R2.4)**: Evaluates oxidation risk (fat > 15%), hygroscopic bounds, fresh produce respiration windows, and acidic liquid constraints.
3. **Deterministic Ranking & Audit Engine**: Computes bounded barrier-fit scores and generates structured audit narratives.

```python
import sys
from pathlib import Path
import pandas as pd

# Add src directory to import path
sys.path.insert(0, str(Path.cwd().parent))

from src.rule_engine import RuleEngine, load_food_profiles_from_db, load_materials_from_db, PackagingContext

db_path = Path.cwd().parent / "database" / "app.db"
engine = RuleEngine.from_config()
food_profiles = load_food_profiles_from_db(str(db_path))
materials = load_materials_from_db(str(db_path))

print(f"Loaded {len(food_profiles)} food profiles and {len(materials)} material candidates from SQLite DB.")
```

## 1. Data Quality Audit Report
Run independent dataset quality check before running recommendation queries.

```python
report = engine.data_quality_report(food_profiles, materials)
pd.DataFrame([report]).T.rename(columns={0: "Count"})
```

## 2. High-Fat Food Recommendation (Fried Peanuts / Cashews)
High-fat content (> 15%) triggers rule R2.1 emitting an OTR constraint <= 10.0 cm³/m²/day.

```python
# Find high-fat food item
high_fat_food = next(f for f in food_profiles if f.fat_pct and f.fat_pct > 30.0)
print(f"Selected Food: {high_fat_food.name} (Code: {high_fat_food.code}, Fat: {high_fat_food.fat_pct}%, Category: {high_fat_food.food_category})")

result = engine.recommend(high_fat_food, materials, top_k=5)

print(f"\n--- FEASIBLE CANDIDATES ({len(result.candidates)}) ---")
feasible_data = [{
    "Rank": idx + 1,
    "Material ID": r.material_id,
    "Material Name": r.material_name,
    "Barrier Score": r.score,
    "Status": r.status,
    "Regulatory": r.regulatory_status,
    "Limiting Barrier": r.limiting_barrier_type
} for idx, r in enumerate(result.candidates)]
display(pd.DataFrame(feasible_data))
```

## 3. Rejection & Audit Log Analysis
Examine rejected candidates and inspect explicit rule failure reasons.

```python
rejected_data = []
for r in result.rejected_candidates[:10]:
    failed_rule = next((rr for rr in r.rule_results if rr.status == "FAIL"), None)
    rejected_data.append({
        "Material ID": r.material_id,
        "Material Name": r.material_name,
        "Status": r.status,
        "Failed Rule": failed_rule.rule_id if failed_rule else "Barrier Constraint",
        "Reason": failed_rule.reason if failed_rule else "Permeability threshold exceeded"
    })
display(pd.DataFrame(rejected_data))
```

## 4. Human-Readable Audit Explanation Narrative
Every recommendation produces an auditable narrative detailing activated rules and assumptions.
